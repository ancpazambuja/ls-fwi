```python

from multiprocessing.pool import ThreadPool

def print_lib_versions(verbose=False):
    if verbose:
        print("Python version")
        print (sys.version)
        print("Version info.")
        print (sys.version_info)
        print("")

def f(x):
    return x*x

class myclass:
    def __init__(self, param=1): 
        self.param = param
    
    def __call__(self, task):
        verbose = os.environ['VERBOSE']
        print(f"param = {self.param}\n")
        if verbose == 'True':
            print(f"task = {task}\n")
        else:
            print(f"Nothing to print... = {task}\n")
        return task['task'][0] + task['task'][1]
    
def myfunc(task):
    verbose = os.environ['VERBOSE']
    if verbose == 'True':
        print(f"task = {task}\n")
    else:
        print(f"Nothing to print... = {task}\n")
    #time.sleep(5)
    return task['task'][0] + task['task'][1]


if __name__ == "__main__":
    print_lib_versions(verbose=True)
    os.environ['VERBOSE'] = 'True'
    obj_instance = myclass(param=2)
    with ThreadPool(torch.cuda.device_count()) as p:
        results = p.map(obj_instance, [
            {"cuda_device" : 0, "task" : (1,2)},
            {"cuda_device" : 1, "task" : (3,4)},
            {"cuda_device" : 2, "task" : (5,6)},
            {"cuda_device" : 3, "task" : (7,8)},
            ])#.apply(myfunc,())
    print(f"After Pool - resultss = {results}") 

#if __name__ == '__main__':
#    with Pool(5) as p:
#        print(p.map(f, [1, 2, 3]))
```
